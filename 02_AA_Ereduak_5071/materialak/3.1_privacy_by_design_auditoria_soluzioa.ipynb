# Ariketa 3.1 — Privacy by Design auditoria — SOLUZIOA

- **Denbora**: 30 min
- **Modalitatea**: taldeka

**Notebook honek soluzio osoa du. Irakasleentzat soilik.**

## 1. Sistemak normalean biltzen dituen datuak (PbD gabe) — SOLUZIOA

Hau da merkatuko produktibitate-monitorizazio tresnek (Hubstaff, Time Doctor, Teramind, Microsoft Productivity Score, etab.) biltzen dutenaren laburpena:

1. **Saio-hasiera/itxiera** (ordutegia, atsedenak).
2. **Aplikazio aktiboen denbora-erregistroa** (zer app irekita dagoen, zenbat denbora `focused`).
3. **Teklatu-zaratak** (*keystrokes per minute*, sarritan editor-mailan).
4. **Saguaren mugimendua eta klikak**.
5. **Ireki diren URL/dokumentuak** (nabigatzaileko historia lan-orduetan).
6. **Pantailaren screenshot periodikoak** (5-15 minutuz behin sarritan).
7. **Webcam-bidezko attention monitoring** (langilea pantailara begira ote dagoen).
8. **Posta-zerbitzariko meta-datuak** (norekin idatzi duen, ordua, baita gaia ere).
9. **Bideo-deien iraupena eta partaideak** (Zoom/Teams logetatik).
10. **Geolokalizazioa** (BYOD edo enpresa-portatilen kasuan).
11. **Imprimatze- eta USB-deskarga-aktibitatea** (DLP — *data loss prevention* gisa marketing-aturik).

Hau guztia **langileari atxikitako identifikadore batekin** (user ID + email + IP). Hau da, **datu pertsonal guztiak GDPR-ren zentzuan**.

**Argibide kritikoa**: kategoria batzuk kategoria sentikorrak izan daitezke (webcam = biometria → GDPR 9. art.; geolokalizazioa = bizilekuari buruzko inferentzia).

## 2. Cavoukian-en 7 printzipioen hausteak — SOLUZIOA

| # | Printzipioa | Hausten al du? | Zergatik |
|---|-------------|----------------|----------|
| 1 | Proaktiboa | **BAI** | Arriskuak ez dira *a priori* aztertu (langile-eskubideen gainean monitorizazio masiboa). EIPD-rik ez normalean. |
| 2 | Privacy as Default | **BAI** | Lehenetsia "dena bildu" da, ez "minimoa". Langileak EZ du opt-in egin behar; opt-out ere ez normalean. |
| 3 | Diseinu integratua | **BAI** | Produktibitate-helburua arrazoi nagusi gisa; pribatutasuna geroago saiatu — ez integratu hasieratik. |
| 4 | Betebehar osoa | **BAI** (alderantziz) | Pribatasuna eta funtzionalitatea aurkakoak balira bezala tratatzen ditu: monitorizazio gehiago = "produktibitate hobea" (frogatu gabea, gainera). |
| 5 | Bizi-ziklo osoko babesa | **BAI** | Log-ak mugagabe gordetzen dira maiz, ezabatze-politika argirik gabe. Datuak inoiz ez dira benetan ezabatzen. |
| 6 | Gardentasuna | **BAI** | Langileek ez dakite zer biltzen den eta nola (politika ezkutua kontratuetan, edo "sistema-monitorizazio orokorra" formulazio lausoa). |
| 7 | Erabiltzaile-errespetua | **BAI** | Erabakia enpresari atxikia, ez langileari. Ahanztura-eskaeretarako mekanismorik ez. |

**Ondorioa**: sistema honek **7 printzipio guztiak hausten ditu**. Hau ez da "design choice tartekoa"; **diseinu-akats sistemikoa** da.

Gainera, urratzen ditu:

- **GDPR 5.1.c (datu-minimizazioa)**: "behar adina, ez gehiago" → biltzen den gehiena ez da beharrezkoa.
- **GDPR 88. art. (lan-testuingurua)**: estatu kideek lan-harremanetan datu-tratamendurako arau espezifikoak ezar ditzakete — Espainian, langileen ordezkariekin akordioa beharrezkoa da.
- **AI Act III. eranskina, 4. atala**: enplegua eta langileen kudeaketa **arrisku altuko erabilera**.

## 3. Sistema berridatzia PbD-rekin — SOLUZIOA

**A) Helburu zehatza eta proportzionatua:**

Ez "produktibitatea" orokorrean, baizik **"talde-mailako kargaren orekatzea, hilabetero"** — adib. departamentu batek karga handiegia duen ikusi eta gizakiak gehitu/birbanatu.

**B) Datu-minimizazioa eta agregazio-maila:**

Norbanako-mailako monitorizazioaren ordez, **talde-mailako agregatua**:

- **Ez** langile zehatzaren denbora
- **Bai** proiektu/talde-arteko karga agregatua (≥5 langileko taldeak agregaziorako).
- Datu indibidualak inoiz behar badira (adib. babes-egoera batean), **seudonimizatuak** eta esplizituki epe motzean (adib. 30 egun) ezabatuak.

**C) Privacy by Default:**

- Langile berri batek lehenetsi sistemarako **parte-hartze minimoa** du (ordu-erregistro hutsa).
- Gehiago jakitearen aukera **opt-in** da (langileak esplizituki onartu behar du), ez opt-out.
- Webcam, screenshot, keystroke: lehenetsi **desaktibatuak**.

**D) Bizi-zikloa:**

- Log-en biltegiratze-epea: gehienez 30-90 egun.
- Ondoren: agregatzea edo ezabatzea automatikoki (cron job auditatua).
- Salbuespena: legezko ikerketa irekia (segurtasun-intzidentea) — orduan ere epe mugatua.

**E) Gardentasuna:**

- Politika idatzia eskuragarri intranetean.
- Langile-batzordearen kontsulta eta akordioa (Langileen Estatutua 64. art.).
- Dashboardak langileari berari ikusgai: nork ikusi ditu nire datuak eta noiz.
- AI Act 50. art.: langileari jakinarazi AA sistema bat erabiltzen ari dela ebaluazio-prozesuetan.

**F) Sarbide-kontrola:**

- HR Director-ek soilik (ez team-leader zuzenak) datu agregatuak ikus ditzakete.
- Datu indibidualetara sarbidea (kasu zehatzetan) auditatua + bigarren begi-onarpenarekin.
- Audit trail iraunkorra: nork sartu den, noiz, zergatik.

**G) Osagai teknikoak:**

- **Seudonimizazioa**: langile-ID → hash kriptografikoa (rotated).
- **Agregazioa**: talde-mailako estatistikak baino ez argitaratu (≥5 langile minimo).
- **Pribatasun Diferentziala**: ε zuhur batekin (adib. ε ≤ 1) enpresak joerak ikus ditzake langile zehatzak identifikatu gabe.
- **EIPD (GDPR 35. art.)** sistema zabaldu aurretik.
- **Enkriptazioa**: atsedenean eta transmisioan, key rotation politikoarekin.

## 4. %100ean bildu BEHAR EZ litzatekeena — SOLUZIOA

| Datu-mota debekatua | Arrazoia |
|---------------------|----------|
| **Webcam edo screenshot periodikoak** | *Intrusibotasun handiena* (langilearen aurpegia/pantailaren gainean); funtzionalitatera ekarpen txikia; alternatiba sinpleagoak daude (zeregin-amaiera baieztatzea). Espainiako Konstituzio Auzitegiak (STC 39/2016) dagoeneko mugatu du. |
| **Keystroke-edukia** | Zer idazten den ezagutzea (ez bakarrik tasa) **komunikazio-pribatuetara** sartzea da (lan-bermean datuak eta posta pertsonala nahas daitezke). Birus baten parekoa. |
| **URL pribatuen historia lan-orduetatik kanpo** (BYOD kasuetan) | Lan-orduetatik kanpoko aktibitatea ez dago enpresaren legezko interes-eremuan. GDPR 6. art. (oinarri juridikoa) faltan. |
| **Posta-edukiak** | Subjektua eta edukia komunikazio-isilpekoaren parte dira (Konstituzioaren 18.3 art. Espainian). Metadatuak baino ez (ordua, hartzailea), gainera proportzionalki neurtuta. |
| **Biometria** (pultsazioa, jarrera, fatiga-detekzioa kamarekin) | Ez bada beharrezkoa segurtasun fisikorako (sektore arriskutsuak: meatzaritza, energia) **GDPR 9. art.** dela eta bereizita tratatu behar; gehienetan, proportzionaltasuna gainditzen du. |
| **Langilearen identifikadorea agregatu gabeko datuetan publikoki erakustea** | Ranking publikoak ("langile-mailako produktibitate-tauletan") **kalte erreputazionala** sortzen du, eta diskriminaziorako bide ireki. |

## 5. Lege-erreferentziak — SOLUZIOA

1. **GDPR 5.1.c** — datu-minimizazioa: "datuak helburuari egokituak, esanguratsuak eta beharrezkoak".
2. **GDPR 6. art.** — oinarri juridikoa: zein oinarri juridikoak justifikatzen du? Konponbidean, ez baimena bakarrik (langile-enpresa harremanaren desorekagatik), baizik kontratuaren beharra **proportzionaltasunarekin**.
3. **GDPR 9. art.** — kategoria sentikorrak (biometria webcam-ekin sortzen denean).
4. **GDPR 22. art.** — erabaki automatizatu indibidualizatu eskubidea — "produktibitate-puntuazio" automatikoa horretarako sartzen da.
5. **GDPR 25. art.** — Privacy by Design eta by Default.
6. **GDPR 35. art.** — EIPD derrigorrezkoa monitorizazio sistematikoetan.
7. **GDPR 88. art.** — estatu kideek lan-testuinguruan arau espezifikoak.
8. **Espainiako Langileen Estatutua 20.3 art.** — enpresariaren langileen kontrol-eskubidea, baina **duintasuna errespetatuz**.
9. **LO 3/2018 (Espainia), 88. art.** — langile-pribatasuna eskubidea digital-testuinguruan.
10. **AI Act III. eranskina, 4. atala** — enplegua = arrisku altuko AA erabilera.
11. **AI Act 14. art.** — giza gainbegiraketa erabaki automatizatuetan.
12. **AI Act 50. art.** — gardentasuna AA sistemen erabileran erabiltzaileei (langileak).

## 6. Hausnarketa — SOLUZIOA

**Kostua PbD-rekin:**

Egia da inplementazio-kostua **handiagoa** dela hasieran (EIPD-a, arkitektura-berrikuspena, langile-batzordearekin kontsulta). Baina:

- **Erregelamendu-zigorrak ekiditen ditu**: GDPR-pean 20 milioi € edo negozioaren %4 arteko zigorrak. AEPD-k 2023an enpresa bat zigortu zuen 6 milioi €-rekin produktibitate-monitorizazio inbaditzaileagatik (legezko jurisprudentzia oso aldakorra da, baina norabidea PbD-aren alde dago).
- **Langile-konfiantza handitu**: monitorizazio inbaditzaileak langile-rotazioa handitzen du.
- **Berrikuntza-onura**: PbD planteatzeak bestelako galderak egitera behartzen du (zer da produktibitatea benetan?).

**Industria erregulatuetan (nuklear/farmazia):**

Sektore erregulatuetan, monitorizazio gehigarria **proportzionatua** izan daiteke, baina:

- Mugatuta segurtasun-helburuetara (ez produktibitatea).
- Datu sentikorrak (biometria, audio): legezko oinarri zehatzarekin (sektore-arau espezifikoa), ez baimen orokorrarekin.
- Atxikipen-epe motza, sarbide-kontrol estua.
- Erregulatzaileari (CSN nuklearrean, AEMPS farmazian) jakinarazpena.
- Langile-batzordearekin akordioa beti.

**Printzipio gakoa**: monitorizazioa helburuari **proportzionatua** izan behar du. Helburu-aldaketa (segurtasun → produktibitate) datu-erabilera-aldaketa onar gabea suposatuko luke (GDPR 5.1.b — helburu-mugaketa).
