## 1. Sailkapena — SOLUZIOA

| Neurria | Kategoria | Helburua | Lege-art. |
|---------|-----------|----------|-----------|
| Entrenamendu-datuak seudonimizatzea | **Privacy by Design** | Datuen subjektua (pertsona) babestu re-identifikaziotik. | GDPR 25. art. (+ 4.5 art. seudonimizazioaren definizioa). |
| Fallback mekanismoa erasopean dagoela | **Security by Design** | Sistemaren erresilientzia: erasoaren aurrean giza-erabakitzailera bideratu. | AI Act 14. art. (giza gainbegiraketa) + 15. art. (sendotasuna). |

| Log sistema: nork kontsultatu duen erregistratzea | **Security by Design** | Auditoria-gaitasuna eta erantzukizuna. | AI Act 12. art. (gertaeren erregistroa). |

**Banaketa:** Privacy 3, Security 4.

## 2. Elkarren osagarritasuna — SOLUZIOA

**A) Zergatik ez da nahikoa Privacy by Design soilik?**

Privacy by Design soilarekin, datuak gutxi dira eta minimo gordeta, baina **sistemak kanpo-erasoak jasaten ditu**. Erasotzaile batek (data poisoning, model leakage) eredua manipulatu edo lapurtu dezake. Dauden datu gutxi horiek ere lapurtu daitezke (poisoning-ek atzerako bidean datu-ihesa eragin dezake), eta sistemaren funtzionamendua bera arrisku larrian dago (diagnostiko okerrak, kreditu-iragazketa ustela). Pribatasuna **bermatuta dago** baina **sistemaren bizia bera kolokan**.

**B) Zergatik ez da nahikoa Security by Design soilik?**

Security by Design soilarekin, sistema sendoa da kanpo-erasoen aurka (filtroak, fallback, log-ak), baina **datu pertsonal gehiegi gordetzen ditu**. Haustura bat gertatzen denean — ezinbestean noizbait gertatuko da, sistema seguruak ere huts egiten du — **inpaktua katastrofikoa** da: milioika erabiltzaileren datu pertsonalak agerian. Ekuifax (2017), Cambridge Analytica (2018), edo Meta-ren etengabeko isunak adibide klasikoak dira: defentsa-arkitektura ona zuten, baina datu-pilaketa masiboak izugarrizko kaltea egin zuen.

**C) Adibide praktikoa: bankuko kreditu-iragazketa**

- **Security by Design (sistema)**: eredura sarrera kontrolatua (rate limiting, IP whitelist), sarreraren iragazketa (prompt injection saihesteko), audit-log osoa, fallback giza analistera <%70 konfiantzan.
- **Privacy by Design (datuak)**: seudonimizazioa entrenamendu-datuetan (izenak ezabatu, ID-a jarri), data minimization (telefono-zenbakia ez biltzea bezeroaren iragazketarako), DP-SGD entrenamenduan (kide-inferentzia saihesteko), ahanztura-eskubidearen mekanismoa (GDPR 17. art.).

**Kasua**: erasotzaileak fallback mekanismoa aktibatu badu (Security geruzak ondo funtzionatu du anomalia detektatuz), heldu duen datu-multzoa **seudonimizatuta dago** eta **diferentzialki babestuta** → kalte erreala minimoa, datu pertsonalak ezin dira erabiltzaile errealekin lotu.

Ondorioa: **defense in depth bi norabidetan** — sistemarako (Security) eta datuetarako (Privacy). Bakar batekin ez da nahikoa GDPR + AI Act betetzeko.

## 3. Hausnarketa: defense in depth bi norabidetan — SOLUZIOA

Security-geruzak hausten badira (kanpoko erasotzailea sartu da), Privacy-geruzek **kalte-mugatzea** bermatzen dute:

- Eraso-aurkari heldu zaion datu-base hori **seudonimizatua** dago → ezin du jakin nor den paziente bakoitza.
- **Pribatasun Diferentziala**-rekin entrenatutako eredutik kide-inferentzia ia ezinezkoa da.
- **Data-minimizazioa** dela eta, hasieratik bildu ez ziren datuak (adib. telefono-zenbaki pribatuak) ezin dira ihesik egin.
- **Atxikipen-epeak**: 30 egunetik gorako log-ak agregatuak dira, beraz, eraso historikoetatik berreskura daiteke datu gutxi.

Hau da **defense in depth-aren** funtsa: **inkonpletua dela onartzea** geruza bakar batek, eta horregatik **gehiago jartzea**. Pribatasuna eta segurtasuna ez dira ordezkagarriak; **multiplikatiboak** dira:

$$\text{Kalte erreala} \approx P(\text{Sec geruzak hausten}) \times P(\text{Pri geruzak hausten}) \times \text{Datu sentikorra}$$

Bi probabilitateak 0.1 baldin badira independenteki, kombinatutako kaltea 0.01 baino ez da; bakarra baldin balitz, 0.1.

## 4. Ebaluatzailearen gida

| Sailkapena | Erantzun zuzena |
|------------|-----------------|
| Seudonimizazioa | Privacy |
| Fallback | Security |
| Data minimization | Privacy |
| Sarrera-iragazketa | Security |
| Pribatasun Diferentziala | Privacy |
| Log sistema | Security |
| API sarbide-kontrola | Security |

**Akats ohikoak ikasleengan:**

- Log sistema → batzuk "Privacy" gisa sailkatzen dute (ez da: helburua **gobernantza/auditoria** da, ez datu-minimizazioa).
- API sarbide-kontrola → gris zertxobait dago, baina **eraso-azalera-minimizazioaren** zentroan dago = Security.
- Seudonimizazioa → batzuk "Security" gisa identifikatzen dute pentsatuz datuak "ezkutatzen" dituela; baina helburua datu-subjektuaren babesa da = Privacy.

**Puntu nagusi pedagogikoa**: bi kategoriek **helburu desberdinak** dituzte, ez teknika ezberdinak. Teknika bera (adib. enkriptazioa) bi mailetan agertu daiteke; sailkapena **zer babesten den**-en araberakoa da.
