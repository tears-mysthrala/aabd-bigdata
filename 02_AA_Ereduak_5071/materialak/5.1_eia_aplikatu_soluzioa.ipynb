## 1. AI Act sailkapena — SOLUZIOA

**A) Sistema arrisku altukoa al da?**

Bai. Bide-trafiko-sistema bat **pertsonak identifikatzen** dituena espazio publikoan eta erabaki/aurreikuspenak hartzen dituena **azpiegitura kritikoan** (errepidea):

- **AI Act III. eranskina, 2. atala**: *Azpiegitura kritikoen kudeaketa eta funtzionamendua* — bide-trafikoa, ur-hornidura, elektrizitatea, etab.
- **AI Act III. eranskina, 1. atala**: biometria erabiltzen bada **denbora errealean urruneko identifikazio biometriko** gisa (*real-time remote biometric identification*).

**B) Salbuespen-galdera — debekuaren arriskua:**

AI Act 5. art.-ek **debekatzen** ditu denbora errealeko urruneko identifikazio biometriko sistemak espazio publikoan, **salbuespen estuetatik kanpo** (lapur larrien bilaketa zehatza, terrorismo-mehatxua, NDA-ko mandamendu judizialarekin). Trafiko-kudeaketa **EZ** dago salbuespenen artean, beraz, aurpegi-identifikaziorik egiten bada, sistema debekatua egon liteke.

Aukerak:

- **Edo** aurpegi-identifikazioa kentzea (matrikulei eta ibilgailuei mugatzea).
- **Edo** sistema-aldaketa: aurpegi-blurring kamerean bertan, identifikazio biometrikoa ez egiteko.

**C) Derrigorrezko artikuluak:**

Arrisku altuko sistema gisa, hauek derrigorrezkoak dira:

- **9. art.** — arrisku-kudeaketa sistema.
- **10. art.** — entrenamendu-datu-multzoen gobernantza (kalitatea, alborapena).
- **12. art.** — gertaeren erregistro automatikoa (audit-log).
- **13. art.** — gardentasuna eta informazioa erabiltzaileei.
- **14. art.** — giza gainbegiraketa.
- **15. art.** — zehaztasuna, sendotasuna, zibersegurtasuna.
- **III. eranskina**, 1. atala (biometria, baldin balitz) + 2. atala (azpiegitura kritikoa).
- **5. art.** — debekuen egiaztapena.

Eta GDPR-tik:
- **9. art.** — datu biometrikoak kategoria sentikorra.
- **25. art.** — Privacy by Design.
- **35. art.** — EIPD derrigorrezkoa.

## 2. EIA-Segurtasuna — SOLUZIOA

| Galdera | Erantzuna | ❌ bada, neurria |
|---------|-----------|------------------|
| Estres-probak (adversarial testing) egin dira? | **❌** sarritan udal-proiektuetan. | Red-team teknikoa kontratatu STOP/seinale-perturbazioak, sticker-erasoak, kamera-itsutzeak (laser, argi infragorria) probatzeko zabaldu aurretik. |
| Detekzio automatikoa manipulazioarekiko? | **❌** ohikoa da. | Anomaly-detection geruza bat (autoencoder) kameren input bideoan, eta alarma SOC-rantz (*Security Operations Center*). |
| Fail-safe protokoloa? | **❌** maiz ("sistema beti aktibo"). | AI ziurtasuna `< X%` denean, semaforo-segidari **arau-oinarriko logika tradizionalera** itzultzeko fallback (AI Act 14. art.). |
| Eraso-azalera murriztua? | **Partzialki** — kamera-API publikoa izan daiteke. | Kameren API-ak barne-sare itxian; segmentazio-firewall-a; ez sare publikora irekita. |
| Audit-trail iraunkorra? | **Partzialki** — udalak `event-log` partikular bat du, baina ez beti tamper-proof. | Immutable audit-log (write-once storage edo blockchain hash chain), GDPR 33. art. eta AI Act 12. art. betetzeko. |

## 3. EIA-Pribatasuna — SOLUZIOA

| Galdera | Erantzuna | ❌ bada, neurria |
|---------|-----------|------------------|
| Datu pertsonal sentibera identifikatuak? | **❌** sarritan ez dago argi. | Datu-mapaketa egitea: matrikulak (PII), aurpegiak (biometria, GDPR 9. art.), oinezkoen ibilbideak (kokapen-datu pertsonal). |
| Privacy by Design teknikak aktibo? | **❌** maiz. | Kameran bertan **edge processing** (matrikula irakurri eta hash gisa transmititu, ez argazkia), **aurpegi-blurring** lehenetsi, datu gordinen biltegiratzea ezabatu (errealdenboran prozesatu eta ezabatu). |
| Ezabatze-eskubidea + herritarrekiko gardentasuna? | **❌** maiz. | Kanpoko kartelak ("Kameraz babestutako gunea — udalaren ardura, info: …"), GDPR 13.-14. art.-aren arabera; web-formulario bat ezabatze/sarbide-eskaerak egiteko; gehienezko biltegiratze-epea (adib. 7-30 egun bideo gordinarentzat). |
| Gardentasuna erabiltzaileei? | **❌** lehenetsi sarritan. | Urteroko gardentasun-txostena udalak argitaratu (zenbat ibilgailu, intzidente, datu-eskaera jaso, etab.). |
| Datu-bizizikloa definitua? | **Partzialki** — sarritan ez espliziki. | Bideo gordina: 7 egun. Metadatu agregatuak: 30-90 egun. Gertaeretarako (poliziak irekitako kasua): legezko epea. Cron job automatizatua ezabaketa-prozesuetarako. |

## 4. Security by Design vs Privacy by Design banaketa — SOLUZIOA

### Security by Design neurriak

1. **Eraso-azalera minimizatzea**: kameren API-ak barne-sare itxian, sare publikora ez irekita.
2. **Defentsa sakona**: enkriptazioa transmisioan + autentifikazioa edge nodoetan + SIEM zentrala.
3. **Gutxieneko pribilegioa**: udal-langileek soilik beren sailaren kamerak ikus ditzakete; ez dute administratzaile-eskubiderik.
4. **Fail-secure**: matxura batean, semaforoak modu seguruenera (laranja kliskaria) pasatzen dira.
5. **Auditoria**: log iraunkorra (*immutable*) — nork eta noiz bideoa ikusi duen.
6. **Kontingenzia-plana**: ereduaren akats bat (drift, eraso) detektatuta, sistema desaktibatu eta giza-operadorea aktibatu.
7. **Adversarial testing (red-team)** zabaldu aurretik eta urtekoa.

### Privacy by Design neurriak

1. **Proaktibo**: EIPD (GDPR 35. art.) eta AESIA-rekin kontsulta proiektua hasi aurretik.
2. **Privacy as Default**: kameraren irteera lehenetsia matrikula hash + ibilgailu-mota; **aurpegirik EZ atera lehenetsiz**; biometria desaktibatu lehenetsi.
3. **Data minimization**: oinezkoen aurpegi-blurring kameran bertan; datuak agregatu sailkatzeko (ibilgailuen kontaketa) banakako jarraipenetik bereizita.
4. **Pribatasun Diferentziala** udal-mailako estatistikak publikatzean (ibilgailu-tasak orduen arabera) ε ≤ 1.
5. **Bizi-ziklo osoa**: bideo gordina ezabatu 7 egun barru (poliziak gertaera bat irekitzen badu izan ezik); meta-datu agregatuak luzeago gorde daitezke.
6. **Erabiltzaile-errespetua**: kartelak, eskubide-formulario digitala, urteroko gardentasun-txostena udalak argitaratu.
7. **Edge processing**: matrikula-irakurketa kameran bertan; ez transmititu argazki gordinik.

## 5. Hausnarketa — SOLUZIOA

**A) Gomendioa:**

**Ez zabaldu sistema egungo egoeran.** Hauek dira arrazoiak:

- AI Act 5. art.-en araberako **debekuaren arriskua** (biometria denbora errealean) — proiektua birdiseinatu behar da aurpegi-identifikazioa kenduz.
- ≥5 ❌ identifikatu ditugu, batzuk kritikoak (fail-safe, edge processing, gardentasuna).
- **EIPD eta AESIA kontsulta** egin gabe daude — GDPR 35. art. eta AI Act 17. art. (kalitate-sistemaren kudeaketa) hausten dira.

**Errekomendazio aktiboa**: birdiseinatu (a) aurpegirik gabe; (b) edge processing matrikula hash-ekin; (c) fail-safe protokoloa eta red-team txostena; (d) EIPD + AESIA aurkezpena. Birdiseinatu ondoren, hilabete batzuetako pilotu-fase mugatua eskualde txiki batean (≤2 km²).

**B) AESIA aurreko EIPD txostenaren hiru elementu nagusi:**

1. **Beharra eta proportzionaltasuna**: zergatik beharrezkoa den sistema (intzidente-tasa baxua, alternatibak agortuta); proportzionaltasun-testa (helburu legitimoa, beharrezkoa eta proportzionatua).
2. **Arriskuen identifikazioa eta mitigazioa**: mehatxu-matrize osoa (lau eraso-bektoreak), Cavoukian-en 7 printzipioak aplikatzeko plana, log-bizizikloa.
3. **Stakeholderren kontsulta eta erabakitze-mekanismoak**: nola jasotzen dira pertsonen eskaerak (sarbide, ezabatze), urteroko berrikuspena, eskualdeko erakundeekin (Ertzaintza, Trafiko-Zuzendaritza) lankidetza eredu.

**C) Alternatibak (proportzionatuagoak):**

- **Induktore magnetikoak**: ibilgailuen kontaketa eta abiadura neurtzen ditu, identifikaziorik gabe. Datu pertsonalik EZ.
- **Bluetooth/WiFi MAC anonimo agregatua**: mugikorretako MAC-ak anonimizatuak (hash + zarata) eta agregatuak — ibilgailu-fluxuen joerak ikus daitezke pertsona zehatzak identifikatu gabe.
- **Floating Car Data (FCD)**: nabigazio-aplikazioek (Google Maps, Waze) ematen dituzten datuak — anonimizazio-maila operadoreak bermatzen du.
- **Pneumatic tubes** edo zoladura-sentsoreak: oso sinpleak, datu pertsonalik ez.

Zergatik proportzionatuagoa?: helburu bera (trafikoaren neurketa eta aurreikuspena) bete daiteke **kamera bidezko identifikazioa erabili gabe**, beraz GDPR 5.1.c eta 25. art.-en araberako proportzionaltasun-testa **gainditzen ez du** kamera-sistemak.
