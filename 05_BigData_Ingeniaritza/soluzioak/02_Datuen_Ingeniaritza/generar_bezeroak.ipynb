```python
with (DATA / "bezeroak_100.csv").open(encoding="utf-8", newline="") as f:
    bezeroak_csv = list(csv.DictReader(f))
with (DATA / "bezeroak_10000.csv").open(encoding="utf-8", newline="") as f:
    bezeroak_10000_csv = list(csv.DictReader(f))
bezeroak_json = json.loads((DATA / "bezeroak.json").read_text(encoding="utf-8"))["bezeroak"]
with (DATA / "salmentak_10000.csv").open(encoding="utf-8", newline="") as f:
    salmentak_csv = list(csv.DictReader(f))
salmentak_json = json.loads((DATA / "salmentak_10000.json").read_text(encoding="utf-8"))["salmentak"]

assert len(bezeroak_csv) == len(bezeroak_json) == 100
assert len(bezeroak_10000_csv) == 10_000
assert len(salmentak_csv) == len(salmentak_json) == 10_000
assert tuple(bezeroak_csv[0]) == BEZERO_ZUTABEAK
assert tuple(salmentak_csv[0]) == SALMENTA_ZUTABEAK
assert 18 <= int(bezeroak_csv[0]["adina"]) <= 80
assert 5 <= float(salmentak_csv[0]["prezioa"]) <= 2_000
assert 1 <= int(salmentak_csv[0]["unitateak"]) <= 10
print(f"bezeroak: {len(bezeroak_csv)} / {len(bezeroak_10000_csv)}; salmentak: {len(salmentak_csv)}")
```
