## 2. Erabaki-Mugak Marrazteko Funtzioa

Sailkatzaile baten portaera ikusteko, laguntza-funtzio bat idatziko dugu. Funtzio honek sare bat (grid) sortuko du gure datuen mugen artean, sareko puntu bakoitzerako klasea iragarriko du eta eremu bakoitza bere kolorearekin margotuko du. Ondoren, gure benetako datuak (puntuak) gainean marraztuko ditu.

```python
def plot_decision_boundaries(clf, X, y, ax, title):
    """
    Sailkatzaile baten erabaki-mugak planoan marrazten dituen funtzioa.
    """
    # Sarearen mugak definitu (datuen minimoak eta maximoak apur bat zabalduta)
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    
    # Sareko puntuen sareta (grid) bat sortu np.meshgrid erabiliz
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))
    
    # Sareko puntu bakoitzerako eredua erabiliz klasea iragarri
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    # Erabaki-eremuak kolore ezberdinekin margotu
    ax.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.RdYlBu)
    
    # Benetako laginak (loreak) marraztu puntu moduan
    scatter = ax.scatter(X[:, 0], X[:, 1], c=y, edgecolor='k', s=50, cmap=plt.cm.RdYlBu)
    
    # Grafikoaren izenburuak eta etiketak ezarri
    ax.set_title(title, fontsize=14, pad=10)
    ax.set_xlabel('Sepal Length (cm)', fontsize=11)
    ax.set_ylabel('Sepal Width (cm)', fontsize=11)
    
    # Kondaira (lore klaseen izenak) gehitu
    handles, labels = scatter.legend_elements()
    legend_labels = ['Setosa', 'Versicolor', 'Virginica']
    ax.legend(handles, legend_labels, loc="upper right", title="Klaseak")
```
